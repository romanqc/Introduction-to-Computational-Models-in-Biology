**(b) What was the purpose of the model - that is, what was accomplished by building and using the model?**

The authors want a decentralized way for wireless nodes to agree on a common clock. They start from the Mirollo-Strogatz pulse-coupled oscillator model of firefly flashing, then modify it so it still converges when transmission and decoding delays are present. The model is used to show that simple local rules (wait, transmit, listen) can produce synchrony without a central coordinator, and Monte Carlo simulations measure how often and how fast this happens.

**(c) What are the parameters (fixed constants that the modeler chooses) of the model? What are the variables that change over time and/or change as a result of these parameters (sometimes called state varaibles)?**

**Parameters:** the period $T$; the phase threshold $\phi_{th}$ (normalized to 1); the coupling factors $\alpha$ and $\beta$ (set by the dissipation factor $b$ and the amplitude increment $\epsilon$); the transmission delay $T_{Tx}$; the decoding delay $T_{dec}$; the refractory time $T_{refr}$; the waiting time $T_{wait} = T - (T_{Tx} + T_{dec})$; and the number of nodes $N$.

**State variables:** the phase $\phi_i(t)$ of each oscillator, which increases linearly from 0 to $\phi_{th}$ with $\dot\phi_i = \phi_{th}/T$, and each node's current discrete state (wait, transmit, refr, or listen).

**(d) Identify one of the models simplifying assumptions - some known aspects of the real world that the model omits or simplifies.**

I think propagation delays are treated as negligible (about 0.17 µs at 50 m), and all nodes are assumed to have identical clocks with no jitter, even though real fireflies show roughly ±90 ms of variation. Noise and channel gain are also ignored.
